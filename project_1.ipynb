## Step 1: Set up pandas and load the data


### Import pandas:


```python
import pandas as pd
```

### Load the dataset


```python
df = pd.read_csv("UCDP_Nigeria_2007_2018.csv")
df.head()
```

## Step 2: Calculate the mean, median, and mode


### 1. Using pandas


```python
mean_pandas = df["best"].mean()
median_pandas = df["best"].median()
mode_pandas = df["best"].mode()[0]

print("Mean:", mean_pandas)
print("Median:", median_pandas)
print("Mode:", mode_pandas)
```

### 2. The hard way (using only the Python standard library)

I will read the CSV file again without using pandas. Then I will calculate the mean, median, and mode using basic Python operations.


```python
import csv

values = []

with open("UCDP_Nigeria_2007_2018.csv", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        values.append(float(row["best"]))

# Mean
total = 0
for value in values:
    total = total + value
mean_hard = total / len(values)

# Median
values.sort()
middle = len(values) // 2
if len(values) % 2 == 0:
    median_hard = (values[middle - 1] + values[middle]) / 2
else:
    median_hard = values[middle]

# Mode
counts = {}
for value in values:
    if value in counts:
        counts[value] = counts[value] + 1
    else:
        counts[value] = 1

mode_hard = max(counts, key=counts.get)

print("Mean:", mean_hard)
print("Median:", median_hard)
print("Mode:", mode_hard)
```

## Step 3: Data visualization

The chart below shows the number of conflict events recorded in each year. Each `*` represents approximately 1,000 events. The chart uses only the Python standard library.


```python
year_counts = {}

with open("UCDP_Nigeria_2007_2018.csv", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        year = int(row["year"])
        if year in year_counts:
            year_counts[year] = year_counts[year] + 1
        else:
            year_counts[year] = 1

print("Conflict events by year (* = approximately 1,000 events)")
for year in sorted(year_counts):
    stars = round(year_counts[year] / 1000)
    print(year, ":", "*" * stars, year_counts[year])
```
