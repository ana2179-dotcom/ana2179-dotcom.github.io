## Step 2: Calculate the mean, median, and mode


### 1. Using pandas

I use the `best` column, which represents the estimated number of fatalities for each recorded conflict event. Missing or nonnumeric values are excluded.


```python
mean_pandas = df["best"].mean()
median_pandas = df["best"].median()
mode_pandas = df["best"].mode()

print("Mean:", mean_pandas)
print("Median:", median_pandas)
print("Mode(s):", mode_pandas)
```

## Step 3: Data visualization

The following text-based bar chart shows the number of recorded conflict events per year. The data are grouped using pandas, but the chart itself is drawn using only standard Python string operations and `print()`. Each `#` represents an approximately equal number of events, as indicated by the scale.


```python
# Group the events by year (pandas is allowed for the calculations).
if "year" not in df.columns:
    raise ValueError("The dataset does not contain a 'year' column.")

annual_counts = df.groupby("year").size().sort_index()

# Keep the bars at a readable width.
max_bar_width = 40
scale = max(1, (int(annual_counts.max()) + max_bar_width - 1) // max_bar_width)

print("UCDP GED: Recorded conflict events by year")
print(f"Each # represents approximately {scale:,} events.\n")
for year, count in annual_counts.items():
    bar_length = round(count / scale)
    bar = "#" * bar_length
    print(f"{year}: {bar} ({count:,})")
```
